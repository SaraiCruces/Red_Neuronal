```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
```

Cargar datos

```python
train = pd.read_excel("train.xlsx")
test = pd.read_excel("test.xlsx")
sample = pd.read_excel("sample_submission.xlsx")
```

```python
GASTOS = ["RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck"]
```

```python

def crear_features(df):
    df = df.copy()
    # PassengerId -> grupo y tamano del grupo
    df["Grupo"] = df["PassengerId"].str.split("_").str[0]
    df["TamGrupo"] = df.groupby("Grupo")["PassengerId"].transform("count")
    # Cabin -> cubierta / numero / lado
    cab = df["Cabin"].str.split("/", expand=True)
    df["Deck"] = cab[0]
    df["CabNum"] = pd.to_numeric(cab[1], errors="coerce")
    df["Side"] = cab[2]
    # Gastos
    df["GastoTotal"] = df[GASTOS].sum(axis=1)
    df["SinGasto"] = (df["GastoTotal"] == 0).astype(int)
    # Los gastos son 0 si esta en criosueno
    df.loc[df["CryoSleep"] == 1, GASTOS] = df.loc[df["CryoSleep"] == 1, GASTOS].fillna(0)
    for c in GASTOS + ["GastoTotal"]:
        df[c + "_log"] = np.log1p(df[c])
    return df.drop(columns=["PassengerId", "Cabin", "Name", "Grupo"] + GASTOS + ["GastoTotal"])
```

```python
X = crear_features(train.drop(columns="Transported"))
y = train["Transported"].astype(int).values
X_test = crear_features(test)
```

```python
num_cols = ["Age", "CabNum", "TamGrupo"] + [c + "_log" for c in GASTOS + ["GastoTotal"]]
cat_cols = ["HomePlanet", "Destination", "Deck", "Side", "CryoSleep", "VIP"]
```

```python
for d in (X, X_test):
    for c in ["CryoSleep", "VIP"]:
        d[c] = d[c].astype("object").where(d[c].notna(), np.nan).astype(str)

prep = ColumnTransformer([
    ("num", Pipeline([("imp", SimpleImputer(strategy="median")),
                      ("sc", StandardScaler())]), num_cols),
    ("cat", Pipeline([("imp", SimpleImputer(strategy="most_frequent")),
                      ("oh", OneHotEncoder(handle_unknown="ignore"))]), cat_cols),
])

X_tr, X_val, y_tr, y_val = train_test_split(X, y, test_size=0.2, stratify=y, random_state=SEED)
X_tr = prep.fit_transform(X_tr).astype("float32")
X_val = prep.transform(X_val).astype("float32")
X_te = prep.transform(X_test).astype("float32")
print("Dimensiones:", X_tr.shape, X_val.shape, X_te.shape)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X_tr.shape[1],)),
    tf.keras.layers.Dense(64, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(1e-4)),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(32, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(1e-4)),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(1, activation="sigmoid"),
])
model.compile(optimizer=tf.keras.optimizers.Adam(1e-3),
              loss="binary_crossentropy", metrics=["accuracy"])
model.summary()
```

```python
callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor="val_loss", patience=15, restore_best_weights=True),
    tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=5),
]
hist = model.fit(X_tr, y_tr, validation_data=(X_val, y_val),
                 epochs=200, batch_size=64, callbacks=callbacks, verbose=2)
```

```python
loss, acc = model.evaluate(X_val, y_val, verbose=0)
print(f"Accuracy en validacion: {acc:.4f}")

import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(hist.history["loss"], label="train"); ax[0].plot(hist.history["val_loss"], label="val")
ax[0].set_title("Loss"); ax[0].legend()
ax[1].plot(hist.history["accuracy"], label="train"); ax[1].plot(hist.history["val_accuracy"], label="val")
ax[1].set_title("Accuracy"); ax[1].legend()
plt.tight_layout(); plt.savefig("curvas_entrenamiento.png", dpi=120)
```

```python
pred = (model.predict(X_te).ravel() > 0.5)
sub = sample.copy()
sub["PassengerId"] = test["PassengerId"].values
sub["Transported"] = pred
sub.to_csv(f"{RUTA}\\submission.csv", index=False)   # formato Kaggle (True/False)
sub.to_excel(f"{RUTA}\\submission.xlsx", index=False)  # por si piden xlsx como el sample
print(sub.head())
```
